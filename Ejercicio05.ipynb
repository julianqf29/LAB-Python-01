```python

import random

# =============================================================================
# EJERCICIO 5.1 - Juego de adivinanza
# =============================================================================
print("--- 5.1 Juego de adivinanza ---")

victorias = 0
derrotas = 0
jugar_nuevamente = "s"

while jugar_nuevamente.lower() == "s":
    numero_secreto = random.randint(1, 100)
    intentos = 0
    max_intentos = 7
    adivinado = False
    historial = []

    print("\nHe pensado un número entre 1 y 100. Tienes 7 intentos.")

    while intentos < max_intentos and not adivinado:
        try:
            intento_usuario = int(input(f"Intento {intentos + 1}/7. Ingresa tu número: "))
            historial.append(intento_usuario)
            intentos += 1

            if intento_usuario == numero_secreto:
                print("¡Felicidades! Has adivinado el número.")
                adivinado = True
                victorias += 1
            elif intento_usuario < numero_secreto:
                print("El número secreto es MAYOR.")
            else:
                print("El número secreto es MENOR.")
        except ValueError:
            print("Por favor, ingresa un número entero válido.")

    if not adivinado:
        print(f"Te has quedado sin intentos. El número secreto era: {numero_secreto}")
        derrotas += 1

    print(f"Tu historial de intentos fue: {historial}")
    print(f"Marcador -> Ganadas: {victorias} | Perdidas: {derrotas}")

    jugar_nuevamente = input("¿Deseas jugar otra vez? (s/n): ")

# Explicación: ¿Por qué 7 intentos son suficientes?
# En cada intento (mayor/menor) descartamos la mitad de las opciones restantes.
# Esto es una búsqueda binaria. La cantidad máxima de intentos para 100 opciones
# es el logaritmo en base 2 de 100, redondeado hacia arriba: log2(100) ≈ 6.64.
# Por lo tanto, con 7 intentos siempre se puede garantizar encontrar el número
# si se juega de forma óptima.


# =============================================================================
# EJERCICIO 5.2 - Juego de la vida (Conway)
# =============================================================================
print("\n--- 5.2 Juego de la vida (Conway) ---")

TAMANO = 10

# Inicializar tablero 10x10 de forma imperativa (sin comprensión de listas)
tablero = []
for i in range(TAMANO):
    fila = []
    for j in range(TAMANO):
        fila.append(0)
    tablero.append(fila)

# Patrón inicial (un "planeador" o glider para que se vea el movimiento)
tablero[1][2] = 1
tablero[2][3] = 1
tablero[3][1] = 1
tablero[3][2] = 1
tablero[3][3] = 1

def contar_vecinos_vivos(tablero, fila, col):
    vivos = 0
    # Recorrer el vecindario 3x3
    for df in range(-1, 2):
        for dc in range(-1, 2):
            if df == 0 and dc == 0:
                continue # Ignorar la celda central

            nf = fila + df
            nc = col + dc

            # Cuidar los bordes (que no se salgan de la matriz)
            if 0 <= nf < TAMANO and 0 <= nc < TAMANO:
                if tablero[nf][nc] == 1:
                    vivos += 1
    return vivos

def mostrar_tablero(tablero):
    for fila in tablero:
        linea = ""
        for celda in fila:
            if celda == 1:
                linea += "■ "
            else:
                linea += "· "
        print(linea)
    print()

print("Generación 0 (Inicial):")
mostrar_tablero(tablero)

generacion_actual = tablero

for gen in range(1, 6):
    # Crear un NUEVO tablero para no modificar el actual mientras se lee
    nuevo_tablero = []
    for i in range(TAMANO):
        fila_nueva = []
        for j in range(TAMANO):
            fila_nueva.append(0)
        nuevo_tablero.append(fila_nueva)

    # Calcular la siguiente generación
    for f in range(TAMANO):
        for c in range(TAMANO):
            vecinos = contar_vecinos_vivos(generacion_actual, f, c)
            estado_actual = generacion_actual[f][c]

            if estado_actual == 1:
                # Regla 1: Celda viva con 2 o 3 vecinos sobrevive
                if vecinos == 2 or vecinos == 3:
                    nuevo_tablero[f][c] = 1
                else:
                    nuevo_tablero[f][c] = 0
            else:
                # Regla 2: Celda muerta con exactamente 3 vecinos nace
                if vecinos == 3:
                    nuevo_tablero[f][c] = 1

    generacion_actual = nuevo_tablero
    print(f"Generación {gen}:")
    mostrar_tablero(generacion_actual)

# Explicación: ¿Por qué usar un tablero nuevo?
# Porque las reglas de Conway dictan el estado de una célula basándose en sus
# vecinos en el "instante actual". Si modificamos el tablero original sobre la marcha,
# las células que ya hayamos actualizado afectarán el cálculo de las células
# que están a su alrededor, produciendo resultados incorrectos. El tablero nuevo
# guarda el "futuro" mientras leemos el "presente" del tablero original.
```
