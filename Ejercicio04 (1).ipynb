```python
# EJERCICIO 4.1 - Alcance global y retorno de valores


# Código original (comentado para evitar que detenga el programa)
# contador_global = 0
# def incrementar():
#     contador_global = contador_global + 1
# incrementar()
# print(contador_global)

# ¿Qué error obtienes? ¿Por qué ocurre?
# R: Se obtiene un UnboundLocalError. Ocurre porque al asignar un valor a
# contador_global dentro de la función, Python asume que es una variable local.
# Al intentar leerla antes de asignarla (en el lado derecho del =), lanza el error.

# Corrección 1: Usando la palabra reservada global
contador_global = 0
def incrementar_global():
    global contador_global
    contador_global = contador_global + 1

incrementar_global()
print("Corrección 1 (usando global):", contador_global)

# Corrección 2: Haciendo que la función reciba el valor y retorne el nuevo valor
def incrementar_retorno(valor):
    return valor + 1

contador_global = incrementar_retorno(contador_global)
print("Corrección 2 (usando retorno):", contador_global)

# ¿Cuál de las dos versiones es más fácil de probar y por qué?
# R: La segunda versión (recibir y retornar) es más fácil de probar.
# Al no depender de un estado global, es una función pura: para cualquier valor
# de entrada, siempre dará el mismo resultado y no generará efectos secundarios
# en otras partes del programa.


# EJERCICIO 4.2 - Mutabilidad y reasignación


def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar_viejo(notas):
    notas = []

def reiniciar_nuevo(notas):
    notas.clear() # o del notas[:]

mis_notas = [3.5, 4.0]

# Predicción antes de ejecutar: Se imprime [3.5, 4.0, 4.8]
# Explicación: agregar_nota modifica el objeto original (mutable) usando append.
# reiniciar_viejo reasigna la variable local 'notas' a una nueva lista vacía,
# pero no afecta a la lista original 'mis_notas' fuera de la función.

print("ID original:", id(mis_notas))
agregar_nota(mis_notas, 4.8)
print("ID después de append:", id(mis_notas)) # Mismo ID, mismo objeto
reiniciar_viejo(mis_notas)
print("ID después de reiniciar_viejo:", id(mis_notas)) # Mismo ID, no se vació
print("Lista tras reiniciar_viejo:", mis_notas)

# Modificando reiniciar para que efectivamente vacíe la lista original
reiniciar_nuevo(mis_notas)
print("ID después de reiniciar_nuevo:", id(mis_notas)) # Mismo ID, pero vacío
print("Lista tras reiniciar_nuevo:", mis_notas)


# EJERCICIO 4.3 - Cajero automático con funciones


cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}

def consultar(cuenta):
    print(f"Titular: {cuenta['titular']} | Saldo actual: ${cuenta['saldo']}")

def depositar(cuenta, monto):
    if monto > 0:
        cuenta["saldo"] += monto
        movimiento = {"tipo": "Depósito", "monto": monto, "saldo_resultante": cuenta["saldo"]}
        cuenta["movimientos"].append(movimiento)
        print(f"Depósito exitoso de ${monto}.")
    else:
        print("El monto a depositar debe ser mayor a 0.")

def retirar(cuenta, monto):
    if monto <= 0:
        print("El monto a retirar debe ser mayor a 0.")
    elif cuenta["saldo"] - monto < 0:
        print("Fondos insuficientes. El saldo no puede quedar en negativo.")
    else:
        cuenta["saldo"] -= monto
        movimiento = {"tipo": "Retiro", "monto": monto, "saldo_resultante": cuenta["saldo"]}
        cuenta["movimientos"].append(movimiento)
        print(f"Retiro exitoso de ${monto}.")

opcion = 0
while opcion != 4:
    print("\n--- MENÚ CAJERO ---")
    print("1. Consultar saldo")
    print("2. Depositar")
    print("3. Retirar")
    print("4. Salir")
    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
        consultar(cuenta)
    elif opcion == 2:
        monto_dep = float(input("Ingrese el monto a depositar: "))
        depositar(cuenta, monto_dep)
    elif opcion == 3:
        monto_ret = float(input("Ingrese el monto a retirar: "))
        retirar(cuenta, monto_ret)
    elif opcion == 4:
        print("\nSaliendo del sistema...")
        print("=== EXTRACTO DE MOVIMIENTOS ===")
        if len(cuenta["movimientos"]) == 0:
            print("No se realizaron movimientos.")
        else:
            print(f"{'Tipo':<12} | {'Monto':<10} | {'Saldo Resultante'}")
            print("-" * 45)
            for mov in cuenta["movimientos"]:
                print(f"{mov['tipo']:<12} | ${mov['monto']:<9.2f} | ${mov['saldo_resultante']:.2f}")
    else:
        print("Opción no válida. Intente de nuevo.")
```
