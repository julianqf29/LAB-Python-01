```python
import json

# EJERCICIO 6.1 - Validación de entradas (Inventario del Ejercicio 2)

productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break

    # Validación del precio con try/except, else y finally
    while True:
        try:
            precio = float(input("Precio: "))
        except ValueError:
            print("Error: El precio debe ser un valor numérico. Intente de nuevo.")
        else:
            # EXPLICACIÓN: El bloque 'else' se ejecuta ÚNICAMENTE si NO ocurrió
            # ninguna excepción en el bloque 'try'. Es decir, si el usuario
            # ingresó un número válido, se ejecuta este bloque.
            if precio >= 0:
                break
            else:
                print("Error: El precio no puede ser negativo.")
        finally:
            # EXPLICACIÓN: El bloque 'finally' se ejecuta SIEMPRE,
            # independientemente de si ocurrió una excepción o no, o si se
            # ejecutó un 'break' dentro del 'try'/'else'. Se usa comúnmente
            # para liberar recursos.
            pass

    # Validación de la cantidad
    while True:
        try:
            cantidad = int(input("Cantidad: "))
            if cantidad < 0:
                print("Error: La cantidad no puede ser negativa.")
            else:
                break
        except ValueError:
            print("Error: La cantidad debe ser un número entero. Intente de nuevo.")

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

print("\nInventario cargado en memoria:")
for p in productos:
    print(p)


# EJERCICIO 6.2 - Persistencia del estado (CSV y JSON)

# PARTE 1: Guardar y leer en CSV
def guardar_csv(lista_productos, nombre_archivo):
    with open(nombre_archivo, "w") as archivo:
        for p in lista_productos:
            # Escribir línea por línea
            archivo.write(f"{p['nombre']},{p['precio']},{p['cantidad']}\n")

def leer_csv(nombre_archivo):
    lista_reconstruida = []
    try:
        with open(nombre_archivo, "r") as archivo:
            for linea in archivo:
                linea_limpia = linea.strip()
                if linea_limpia != "":
                    partes = linea_limpia.split(",")
                    producto = {
                        "nombre": partes[0],
                        "precio": float(partes[1]),
                        "cantidad": int(partes[2])
                    }
                    lista_reconstruida.append(producto)
    except FileNotFoundError:
        print(f"El archivo {nombre_archivo} no existe. Se inicia con inventario vacío.")
        lista_reconstruida = []
    return lista_reconstruida

guardar_csv(productos, "inventario.csv")
productos_recuperados_csv = leer_csv("inventario.csv")
print("\nDatos recuperados desde CSV:")
for p in productos_recuperados_csv:
    print(p)


# PARTE 2: Guardar y leer en JSON
def guardar_json(lista_productos, nombre_archivo):
    with open(nombre_archivo, "w") as archivo:
        json.dump(lista_productos, archivo, indent=4)

def leer_json(nombre_archivo):
    try:
        with open(nombre_archivo, "r") as archivo:
            lista_reconstruida = json.load(archivo)
    except FileNotFoundError:
        print(f"El archivo {nombre_archivo} no existe. Se inicia con inventario vacío.")
        lista_reconstruida = []
    return lista_reconstruida

guardar_json(productos, "inventario.json")
productos_recuperados_json = leer_json("inventario.json")
print("\nDatos recuperados desde JSON:")
for p in productos_recuperados_json:
    print(p)


# Retrospectiva
#
# 1. Invertimos 3 horas cada uno
# 2. Finalizado
# 3. Haber aprendido mas sobre python, por que nos puede ayudar mas adelante en otras materias
# 4. No entender como hacer algunas cosas, investigar y apoyarnos en la IA
# 5. Nos dividimos bien el trabajo, nos comprometemos a hacerlo un poco antes
# 6. Python Software Foundation. (2024). The Python Tutorial. https://docs.python.org/3/tutorial/)
```
