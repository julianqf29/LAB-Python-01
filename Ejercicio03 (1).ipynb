```python
import requests

# EJERCICIO 3.1 - Filtrar usuarios por correo .biz


url_usuarios = "https://jsonplaceholder.typicode.com/users"
response_usuarios = requests.get(url_usuarios)
usuarios = response_usuarios.json()

nombres_biz = []
for u in usuarios:
    correo = u["email"]
    if correo[-4:] == ".biz":
        nombres_biz.append(u["name"])

print("Usuarios encontrados con correo .biz:")
for n in nombres_biz:
    print("-", n)


# EJERCICIO 3.2 - Pokemon API (Imperativo y condicionales)

nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()

url_pokemon = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url_pokemon)

if respuesta.status_code == 200:
    datos = respuesta.json()

    tipos = []
    for t in datos["types"]:
        tipos.append(t["type"]["name"])

    habilidades = []
    for h in datos["abilities"]:
        habilidades.append(h["ability"]["name"])

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades,
        "experiencia_base": datos["base_experience"] # Uso adicional de la API
    }

    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")

    if len(pokemon["tipos"]) > 1:
        print("¡Este es un Pokémon de tipo mixto!")

    if pokemon["peso"] > 100:
        print("¡Este es un Pokémon de gran tamaño!")
    else:
        print("Este es un Pokémon ligero.")

    print("Habilidades:")
    for h in pokemon["habilidades"]:
        print(f"- {h.capitalize()}")

    print(f"Experiencia base que otorga al ser derrotado: {pokemon['experiencia_base']}")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
