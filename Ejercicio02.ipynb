```python

# EJERCICIO 2.1 - Condicionales para conducir

print("--- 2.1 Licencia de conducir ---")
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ").lower()
esta_sancionado = input("¿Está sancionado? (s/n): ").lower()

if edad >= 18 and tiene_licencia == 's' and esta_sancionado == 'n':
    print("Usted PUEDE conducir.")
else:
    print("Usted NO puede conducir.")
    if edad < 18:
        print("- Motivo: Es menor de edad.")
    if tiene_licencia != 's':
        print("- Motivo: No tiene licencia.")
    if esta_sancionado == 's':
        print("- Motivo: Se encuentra sancionado.")



# EJERCICIO 2.2 - Suma de primeros n números, pares y acumulador

print("\n--- 2.2 Suma de números naturales ---")
n = int(input("Ingrese la cantidad de números (n): "))
suma_total = 0
suma_pares = 0

for i in range(1, n + 1):
    suma_total += i
    if i % 2 == 0:
        print(f"Número par encontrado: {i}")
        suma_pares += i

print(f"Suma total de los primeros {n} números: {suma_total}")
print(f"Suma de los números pares: {suma_pares}")



# EJERCICIO 2.3 - Temperaturas mayores al promedio

print("\n--- 2.3 Temperaturas mayores al promedio ---")
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1, 18.0, 27.0]
suma_temperaturas = 0

for t in temperaturas:
    suma_temperaturas += t

promedio = suma_temperaturas / len(temperaturas)
print(f"El promedio es: {promedio:.2f}")

print("Temperaturas por encima del promedio:")
for t in temperaturas:
    if t > promedio:
        print(f"- {t}")


# EJERCICIO 2.4 - Ciclo hacia atrás y programa hasta 0

print("\n--- 2.4 Ciclos ---")
print("Conteo hacia atrás:")
contador = 5
while contador > 0:
    print("Contador:", contador)
    contador -= 1
print("¡Despegue!")

print("\nIngreso de números hasta 0:")
# EXPLICACIÓN DEL ESTADO:
# La variable 'numero' representa el estado del ciclo.
# Su valor inicial nos permite entrar al while, y se actualiza (lee un nuevo valor)
# en cada iteración. La condición 'numero != 0' dicta si el ciclo continúa o termina.
numero = -1
suma_numeros = 0
while numero != 0:
    numero = int(input("Ingrese un número (0 para salir): "))
    suma_numeros += numero
print(f"La suma de los números ingresados es: {suma_numeros}")



# EJERCICIO 2.5 - Acumular pares e impares, estado final en diccionario

print("\n--- 2.5 Suma de pares e impares ---")
suma_pares = 0
suma_impares = 0

# EXPLICACIÓN DEL ESTADO DENTRO DEL BUCLE:
# En cada iteración, 'i' toma un valor. Dependiendo de si es par o impar,
# se actualiza la variable suma_pares o suma_impares sumando el valor actual de 'i'.
# El estado de los acumuladores muta en cada vuelta.
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}
print("Estado final:", estado_final)


# EJERCICIO 2.6 - Inventario con continue, max() y lista de diccionarios


print("\n--- 2.6 Inventario de productos ---")
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    if precio < 0 or cantidad < 0:
        print("Error: Precio o cantidad no pueden ser negativos. Producto omitido.")
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})


total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]
print(f"Total del inventario: ${total:.2f}")


def obtener_precio(producto):
    return producto["precio"]

if len(productos) > 0:
    producto_mas_caro = max(productos, key=obtener_precio)
    print("El producto más caro es:", producto_mas_caro)
else:
    print("No se ingresaron productos al inventario.")

print("\nEstado final del inventario (lista de diccionarios):")
for p in productos:
    print(p)
```
